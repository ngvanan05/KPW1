```python
import os
import requests
import re
from bs4 import BeautifulSoup
```

```python
File_Store = 'D:\\10\\Cau1\\Data'

Filename = os.path.join(File_Store, '10_News.txt')
```

```python
url = 'https://hanoimoi.vn/du-lich'

html = requests.get(url).text

soup =  BeautifulSoup(html, 'html.parser')

Tag = soup.find('ul', class_="onecms__loading")


#Tiêu đề
TieuDe = Tag.find_all('h3', class_="b-grid__title")

#Ngày đăng tin
NgayDang = Tag.find_all('span', class_="b-grid__time")

#Tóm tin
TomTin = Tag.find_all('div', class_="b-grid__row b-grid__desc")

#Link Web
html2 = str(Tag)
LinkBaiWeb = re.findall(r'<h3 class="b-grid__title" data-live="0"><a href="(.*?)">.*?</a></h3>', html2)
with open(Filename, 'w', encoding='utf-8') as f:
    for i in range(0,10):
        Tieu_De = TieuDe[i].get_text(strip=True)

        Ngay_Dang = NgayDang[i].get_text(strip=True)
        
        Tom_Tin = TomTin[i].get_text(strip=True)
        
        Link_Bai_Web = LinkBaiWeb[i]    
        
        NoiDung = f'Tiêu đề: {Tieu_De}\nNgày đăng: {Ngay_Dang}\nTóm tin: {Tom_Tin}\nLink bài viết: {Link_Bai_Web}\n'
        
        print(NoiDung)
        
        f.write(NoiDung + '\n') 
```

```python
LinkBaiWeb = re.findall(r'<h3 class="b-grid__title" data-live="0"><a href="(.*?)">.*?</a></h3>', html2)
File_Store2 = 'D:\\10\\Cau1\\Data'
for i in range(0, 10):
    url2 = LinkBaiWeb[i]
    html3 = requests.get(url2).text
    soup2 = BeautifulSoup(html3, 'html.parser')

    # Tựa đề
    Tieu_De_element = soup2.find('h1', class_="sc-longform-header-title block-sc-title")
    TieuDe = Tieu_De_element.get_text(strip=True) if Tieu_De_element else "None"

    # Tác giả
    Tac_Gia_element = soup2.find('span', class_="sc-longform-header-author block-sc-author")
    TacGia = Tac_Gia_element.get_text(strip=True) if Tac_Gia_element else "None"

    # Miêu tả
    description_element = soup2.find('p', class_="sc-longform-header-sapo block-sc-sapo")
    description = description_element.get_text(strip=True) if description_element else "None"

    # Nội dung bài báo
    NoiDungBaiBao_Blocks = soup2.find('div', class_="entry entry-no-padding")
    if NoiDungBaiBao_Blocks:
        lines = NoiDungBaiBao_Blocks.find_all('p')
        NoiDungBaiBao = [description] + [line.get_text(strip=True) for line in lines]
    else:
        NoiDungBaiBao = [description, "None"]

    Content = f'Tựa đề: {TieuDe}\nTác giả: {TacGia}\nNội dung bài báo: ' + ''.join(NoiDungBaiBao)

    print(Content + '\n')

    filename = os.path.join(File_Store2, f'10_news{i+1}.txt')
    with open(filename, 'w', encoding='utf-8') as f:
        f.write(Content + '\n')
    
```
