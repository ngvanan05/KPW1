```python
import os
from collections import Counter

folder_Dulieu = r"D:\\10\\Cau3\\Data\\Data_Cau3"

all_words = []
for file_name in os.listdir(folder_Dulieu):
    if file_name.endswith(".txt"):
        with open(os.path.join(folder_Dulieu, file_name), "r", encoding="utf-8") as f:
            text = f.read().split()
            all_words.extend(text)

freq = Counter(all_words)
print(freq.most_common(20))
```

```python
import matplotlib.pyplot as plt
from wordcloud import WordCloud

#top 30
top30 = freq.most_common(30)
words, counts = zip(*top30)

plt.figure(figsize=(10,5))
plt.bar(words, counts)
plt.xticks(rotation=45)
plt.title("Top 30 từ xuất hiện nhiều nhất")
plt.show()

# Vẽ word cloud
wordcloud = WordCloud(width=1000, height=400, background_color="white").generate(" ".join(all_words))
plt.figure(figsize=(12,6))
plt.imshow(wordcloud, interpolation="bilinear")
plt.title("WordCloud các từ xuất hiện nhiều nhất", fontsize=16)
plt.axis("off")
plt.show()
```
