```python
import re
import os
from bs4 import BeautifulSoup
from underthesea import word_tokenize
```

```python
stopword_path = r"D:\\Kỳ_1_2025_2026\\Khai_Phá_Dữ_Liệu_Web\\Cuoi_Ky_KPW\\10\\Data-20251130T150154Z-1-001\\Data\\Data\\vietnamese-stopwords.txt" 
with open(stopword_path, "r", encoding="utf-8") as f:
    stopwords = set(line.strip() for line in f if line.strip())
    
    print(stopwords)
    
def preprocess_text(text: str): 
    #Loại bỏ thẻ HTML 
    text = BeautifulSoup(text, "html.parser").get_text()

    #Chuyển chữ thường
    text = text.lower()

    # Xóa URL và html
    text = re.sub(r"http\S+|www\S+|https\S+", " ", text)

    # Xóa dấu câu, số và ký tự đặc biệt 
    text = re.sub(r"[^a-zA-ZÀ-ỹ\s]", " ", text)

    # Tách từ tiếng Việt 
    tokens = word_tokenize(text, format="text").split()

    # Loại bỏ stopword 
    tokens = [w for w in tokens if w not in stopwords and len(w) >= 1]

    # Ghép lại thành chuỗi 
    return " ".join(tokens)


input = r"D:\Kỳ_1_2025_2026\Khai_Phá_Dữ_Liệu_Web\Cuoi_Ky_KPW\10\Data-20251130T150154Z-1-001\Data\Data\Cau1\Cau1_ RawData"
output = r"Cuoi_Ky_KPW/10"


index = 1
for file in os.listdir(input):
    if file.endswith(".txt"):
        path_in = os.path.join(input, file)
        with open(path_in, "r", encoding="utf-8") as f:
            content = f.read()
        
        # Tiền xử lý
        result = preprocess_text(content)
        
        # Tạo tên file mới
        file_out = f"10_Pre{index}.txt"
        path_out = os.path.join(output, file_out)
        
        # Ghi ra thư mục đích
        with open(path_out, "w", encoding="utf-8") as out:
            out.write(result)
        print("Xuất File:  ", file_out)
        index += 1      
```
