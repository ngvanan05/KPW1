```python
# Define the dataset
documents = [
    "both good and distinctive",
    "good but not distinctive",
    "distinctive but not good, or",
    "neither good nor distinctive"
]

# Tokenize the documents and create a vocabulary
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer(binary=True)
X = vectorizer.fit_transform(documents)

# Display the One-Hot Encoded vectors
import pandas as pd
one_hot_df = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())
print(one_hot_df)
```

```python
# Define the sentences
sentences = [
    "The quick brown fox jumps over the lazy dog and",
    "Never jump over the lazy dog quickly"
]

# Create a Bag-of-Words (BoW) representation
from sklearn.feature_extraction.text import CountVectorizer

bow_vectorizer = CountVectorizer()
X_bow = bow_vectorizer.fit_transform(sentences)

# Display the BoW vectors
bow_df = pd.DataFrame(X_bow.toarray(), columns=bow_vectorizer.get_feature_names_out())
print(bow_df)
```

```python
# Export the One-Hot Encoding DataFrame to Excel
one_hot_df.index = ['d1', 'd2', 'd3', 'd4']  # Rename rows
one_hot_df.to_excel('10_OneHot.xlsx', index_label='Document')

# Export the Bag-of-Words DataFrame to Excel
bow_df.index = ['d1', 'd2']  # Rename rows
bow_df.to_excel('10_BOW.xlsx', index_label='Sentence')
```

```python
# Sort the Bag-of-Words DataFrame by the first row (d1)
sorted_bow_df = bow_df.sort_values(by='d1', axis=1, ascending=False)
print(sorted_bow_df)

# Export the sorted BoW DataFrame to Excel
sorted_bow_df.to_excel('sorted_BOW.xlsx', index_label='Sentence')
```

```python
# Define the dataset for Bag-of-Words (BoW)
documents_bow = [
    "both good and distinctive",
    "good but not distinctive",
    "distinctive but not good, or",
    "neither good nor distinctive"
]

# Create a Bag-of-Words (BoW) representation
bow_vectorizer = CountVectorizer()
X_bow_documents = bow_vectorizer.fit_transform(documents_bow)

# Display the BoW vectors
bow_documents_df = pd.DataFrame(X_bow_documents.toarray(), columns=bow_vectorizer.get_feature_names_out())
bow_documents_df.index = ['d1', 'd2', 'd3', 'd4']  # Rename rows
print(bow_documents_df)

# Export the BoW DataFrame to Excel
bow_documents_df.to_excel('bow_documents.xlsx', index_label='Document')
```

```python
# Define the dataset for Bag-of-Words (BoW) in Vietnamese
documents_vietnamese = [
    "Cả xuất_sắc và độc_đáo",
    "Xuất_sắc nhưng không độc_đáo",
    "Độc_đáo nhưng không xuất_sắc",
    "Không xuất_sắc và không độc_đáo"
]

# Create a Bag-of-Words (BoW) representation
bow_vectorizer_vietnamese = CountVectorizer()
X_bow_vietnamese = bow_vectorizer_vietnamese.fit_transform(documents_vietnamese)

# Display the BoW vectors
bow_vietnamese_df = pd.DataFrame(X_bow_vietnamese.toarray(), columns=bow_vectorizer_vietnamese.get_feature_names_out())
bow_vietnamese_df.index = ['d1', 'd2', 'd3', 'd4']  # Rename rows
print(bow_vietnamese_df)

# Export the BoW DataFrame to Excel
bow_vietnamese_df.to_excel('bow_vietnamese.xlsx', index_label='Document')
```

```python
# Define the dataset for n-grams
documents_ngrams = [
    "The quick brown fox jumps over the lazy dog and",
    "Never jump over the lazy dog quickly"
]

# Create a feature vector with 1-grams and 2-grams
ngram_vectorizer = CountVectorizer(ngram_range=(1, 2))
X_ngrams = ngram_vectorizer.fit_transform(documents_ngrams)

# Display the n-grams feature vectors
ngrams_df = pd.DataFrame(X_ngrams.toarray(), columns=ngram_vectorizer.get_feature_names_out())
ngrams_df.index = ['Doc1', 'Doc2']  # Rename rows
print(ngrams_df)

# Export the n-grams DataFrame to Excel
ngrams_df.to_excel('ngrams_feature_vectors.xlsx', index_label='Document')
```
