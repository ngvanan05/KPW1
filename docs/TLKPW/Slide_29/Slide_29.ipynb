```python
import requests 
import re
import os
from bs4 import BeautifulSoup
```

```python
url = 'https://phongtro123.com/tinh-thanh/da-nang'
html = requests.get(url).text
```

```python
print(html)
```

```python
TieuDe = re.findall('<h3 class="fs-6 fw-medium mb-2"><a.*?>(.*?)</a></h3>',html)

DonGia = re.findall('<span class="text-green fw-semibold fs-6">(.*?)</span>',html)

Diachi = re.findall('<a class="text-body".*?>(.*?)</a>',html)

Ngaydang = re.findall('<time class="fs-9 text-secondary.*?>(.*?)</time>',html)

DienTich = re.findall(r'<span>(\d+\s*m)<sup>2</sup></span>',html)

DonGia1 = DonGia[1:]
Diachi1 = Diachi[1:]
Ngaydang1 = Ngaydang[1:]
DienTich1 = DienTich[1:]
str = ""
for i in range(len(TieuDe)):
    str +=TieuDe[i] + "\n"
    
    str +="-"+ DonGia1[i] + "\n"
    
    str +="-"+ Diachi1[i] + "\n"
    
    str +="-"+  DienTich1[i] + "\n"
    
    str +="-"+  Ngaydang1[i] + "\n"
print(str)
```

```python
filename=os.path.join("D:\\MyPython\\Slide_29\\PhongTro123.txt")    
with open(filename, 'w',encoding='utf-8') as f:
    f.write(str)
```
