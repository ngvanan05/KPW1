```python
import requests 
import re
import os
from bs4 import BeautifulSoup
```

```python
url = 'https://philong.com.vn/may-tinh-xach-tay.html'
html = requests.get(url).text
soup = BeautifulSoup(html, "html.parser")
```

```python
print(html)
```

```python
Tensanpham = re.findall(r'<h4 class="p-name line-clamp-2">\s*<a.*?>(.*?)</a>\s*</h4>',html)

DonGia = re.findall('<span class="p-price  ">(.*?)</span>',html)

# MÔ TẢ SẢN PHẨM
Mota_blocks = soup.find_all("div",class_="p-summary")

list_mota = []
for block in Mota_blocks:
    Lines = block.find_all('p')
    mota = [line.get_text(strip=True) for line in Lines]
    list_mota.append(mota)

products = []
for ten, gia, mota in zip(Tensanpham, DonGia, list_mota):
    product = {
        "Tên sản phẩm": ten.strip(),
        "Đơn giá": gia.strip(),
        "Mô tả": mota
    }
    products.append(product)

for i, p in enumerate(products, 1):
    print(f"--- Sản phẩm {i} ---")
    print("Tên sản phẩm:", p["Tên sản phẩm"])
    print("Đơn giá:", p["Đơn giá"])
    print("Mô tả:")
    for line in p["Mô tả"]:
        print("-", line)
    print("-" * 50)
      




```
