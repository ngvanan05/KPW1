```python
import requests
from bs4 import BeautifulSoup
import os
```

```python
url = 'https://phongtro123.com/tinh-thanh/da-nang'
html = requests.get(url).text
```

```python
#Tối ưu hóa code HTML bằng thư viện html5lib
soup = BeautifulSoup(html, 'html5lib')
```

```python
#Sử dụng beautifulshop để bóc tách dữ liệu
tieude = soup.find_all("h3",class_ = "fs-6 fw-medium mb-2")

dongia = soup.find_all("span",class_ = "text-green fw-semibold fs-6")

diachi = soup.find_all("div",class_ = "mb-2 d-flex")

ngaydang = soup.find_all("time",class_ = "fs-9 text-secondary")

dientich = [dientich1.find_all("span")[-1] for dientich1 in soup.find_all("div",class_="mb-2 line-clamp-1")]
str = ""

for i in range(len(tieude)):
    str+= tieude[i].text + "\n"
    
    str+= "- "+ dientich[i].text + "\n"
    
    str+= "- " + dongia[i].text + "\n"
    
    str+= "- " + diachi[i].text + "\n"
    
    str+= "- " + ngaydang[i].text + "\n"

    
print(str)
```
