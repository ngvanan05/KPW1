```python
import requests
import re

url = "https://truyenfull.vision/the-loai/kiem-hiep/"

def get_LinksbyRegEx(url_visit):
    global Links_ToDo, Links_seen
    print("** Now visiting:", url_visit)
    
    Links_seen.append(url_visit)    
    html = requests.get(url_visit).text
    NextLinks = re.findall(r'href="(https://truyenfull\.vision/the-loai/kiem-hiep/trang-\d+/)"', html)

    for Link in NextLinks:

        # Chỉ giữ link của truyenfull.vision
        if Link.startswith("https://truyenfull.vision"):
            if Link not in Links_ToDo and Link not in Links_seen:
                Links_ToDo.append(Link)
        
    if Links_ToDo:
        get_LinksbyRegEx(Links_ToDo.pop())
    else:
        return

# Cách 1: Crawl bằng regex + đệ quy
Links_ToDo = [url]
Links_seen = []
get_LinksbyRegEx(Links_ToDo.pop())
print(len(Links_seen), "links found!")
   


#Cách 2: Dựa trên quy tắc cấu trúc url để tạo link
Links_ToDo=[url]
for i in range(2,35):
    Link="https://truyenfull.vision/the-loai/kiem-hiep/trang-" + str(i) + "/"
    Links_ToDo += [Link]

print(Links_ToDo)    
```
