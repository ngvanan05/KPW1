```python
import requests
import re
import os
from urllib.parse import urlsplit

#Tạo thư mục đường dẫn và file url
file_Store = 'D:\\Kỳ_1_2025_2026\\Khai_Phá_Dữ_Liệu_Web\\Slide_43'
url_fileRss = 'https://vnexpress.net/rss/the-thao.rss'

domain = urlsplit(url_fileRss).netloc

# xóa các dấu 
def url_to_file_name(url):
    url = str(url).strip().replace(' ','_')
    return re.sub(r'(?u)[^-\w.]','',url) + '.html'

#Tạo file lưu và file rss
def download(url):
    print("Saving file,...")
    content = requests.get(url).text
    filename = os.path.join(file_Store, url_to_file_name(url))
    
    with open(filename, 'w',encoding='utf-8') as f:
        f.write(content)

def visit(url):
    print("** Now visting:",url)
    download(url)
    
# lấy link
def getUrl_from_filesRSS(url_fileRss):
    xml = requests.get(url_fileRss).text
    links = re.findall('<link>(.*?)</link>',xml)
    
    links_todo = []
    for link_url in links:
        if link_url is None:
            continue
        
        if urlsplit(link_url).netloc != domain:
            continue
        
        if link_url in links_todo:
            continue
        
        links_todo.append(link_url)
    
    return links_todo

# Lấy danh sách link
links_todo = getUrl_from_filesRSS(url_fileRss)
print(links_todo)





        
```
