```python
import re
import os
import requests
from bs4 import BeautifulSoup
from urllib.parse import urlsplit

# Bước 1: Tạo đường dẫn và RSS
file_store = "D:\\Kỳ_1_2025_2026\\Khai_Phá_Dữ_Liệu_Web\\Slide_45"
os.makedirs(file_store, exist_ok=True)

url_fileRss = 'https://thanhnien.vn/rss/kinh-te.rss'

# Bước 2: lấy domain
domain = urlsplit(url_fileRss).netloc

# Bước 3: đổi URL thành tên file
def url_to_file_name(url):
    url = str(url).strip().replace(" ", "_")
    return re.sub(r'(?u)[^-\w.]', '', url) + ".txt"

# Bước 4: Lưu nội dung bài báo
def save_toTXT(url):
    print("Saving file....", url)
    html = requests.get(url).text
    soup = BeautifulSoup(html, "html5lib")
    
    Title = soup.find('h1', class_="detail-title")
    Desc = soup.find("p", class_="description")    
    Detail = soup.find("article", class_="fck_detail")
    
    title_text = Title.text.strip() if Title else ""
    desc_text = Desc.text.strip() if Desc else ""
    detail_text = Detail.text.strip() if Detail else ""
    
    Content = title_text + "\n" + desc_text + "\n" + detail_text
    
    filename = os.path.join(file_store, url_to_file_name(url))
    with open(filename, 'w', encoding='utf-8') as f:
        f.write(Content)

def visit(url):
    print("** Now visiting:", url)
    save_toTXT(url)

# Lấy link từ RSS
def getURL_from_fileRSS(url_fileRSS): 
    xml = requests.get(url_fileRSS).text
    links = re.findall(r'<link>(.*?)</link>', xml)
    links = links[1:]  # bỏ link đầu tiên (link của chính RSS)
    
    links_todo = []
    for link_url in links:
        if not link_url:
            continue
        if urlsplit(link_url).netloc != domain:
            continue
        if link_url in links_todo:
            continue
        links_todo.append(link_url)
    return links_todo

# MAIN
links_todo = getURL_from_fileRSS(url_fileRss)
while links_todo:
    url_to_visit = links_todo.pop()
    visit(url_to_visit)
```
